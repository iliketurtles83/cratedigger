```python
# subset of columns for analysis
features_df = features_df[["year_tag", "genre_tag", "duration_ms", "bpm", "key", "mode", "energy", 
                           "loudness_lufs", "spectral_centroid", "zero_crossing_rate", "danceability", 
                           "speechiness", "acousticness", "instrumentalness", "liveness", "valence"]]
 
#  6   year_tag            107 non-null    str    
#  7   genre_tag           110 non-null    str    
#  8   duration_ms         110 non-null    int64  
#  9   bpm                 110 non-null    float64
#  10  key                 110 non-null    str    
#  11  mode                110 non-null    str    
#  12  energy              110 non-null    float64
#  13  loudness_lufs       110 non-null    float64
#  14  spectral_centroid   110 non-null    float64
#  15  zero_crossing_rate  110 non-null    float64
#  16  danceability        110 non-null    float64
#  17  speechiness         110 non-null    float64
#  18  acousticness        110 non-null    float64
#  19  instrumentalness    110 non-null    float64
#  20  liveness            110 non-null    float64
#  21  valence             110 non-null    float64
#  22  schema_version      110 non-null    int64  
#  23  analyzed_at         110 non-null    str    
#  24  error               0 non-null      object 
```

```python
# categorical features value counts
for col in features_df.select_dtypes(include="object").columns:
    print(f"Value counts for {col}:")
    print(features_df[col].value_counts())
    print("\n")
```

```python
# distribution for numerical features
numerical_cols = features_df.select_dtypes(include="number").columns
features_df[numerical_cols].hist(bins=30, figsize=(15, 10))
```

```python
import seaborn as sns

# correlation heatmap of numeric features
corr = features_df[numerical_cols].corr()

plt.figure(figsize=(12, 10))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Numerical Features")
plt.tight_layout()
plt.show()

```
